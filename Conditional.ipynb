Mini Data Science Task

Use this dictionary:

employee = {
    "Name": "Sai",
    "Salary": 85000,
    "Experience": 6
}

Write three conditional (ternary) expressions that assign values to variables:

salary_status
Salary ≥ 70000 → "High Salary"
Otherwise → "Regular Salary"
experience_level
Experience ≥ 5 → "Experienced"
Otherwise → "Beginner"
bonus
Salary ≥ 80000 → "Bonus Eligible"
Otherwise → "No Bonus"

Finally, print all three variables.

```python
employee = {
    "Name": "Sai",
    "Salary": 85000,
    "Experience": 6
}
salary_status= "High Salary" if employee["Salary"]>=70000 else "Regular Salary"
experience_level="Experienced" if employee["Experience"]>=5 else "Beginner"
bonus="Bonus Eligible" if employee["Salary"]>=80000 else "No Bonus"
print(salary_status)
print(experience_level)
print(bonus)
```
